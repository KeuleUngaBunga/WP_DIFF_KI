```python
import torch
import torch.nn as nn
import torch.nn.functional as F
import math
import numpy as np
import matplotlib.pyplot as plt
from torch.utils.data import DataLoader, Subset
from tqdm import tqdm
import torchvision
import torchvision.transforms as transforms
from torchvision.utils import make_grid, save_image
import copy


# =====================================================================
# TIME EMBEDDING
# =====================================================================
def sinusoidal_time_embedding(t, dim):
    device = t.device
    half = dim // 2
    freqs = torch.exp(
        -math.log(10000) * torch.arange(0, half, device=device) / half
    )
    args = t[:, None] * freqs[None]
    return torch.cat([torch.sin(args), torch.cos(args)], dim=-1)


class TimeMLP(nn.Module):
    def __init__(self, dim):
        super().__init__()
        self.net = nn.Sequential(
            nn.Linear(dim, dim * 4),
            nn.SiLU(),
            nn.Linear(dim * 4, dim)
        )

    def forward(self, t):
        return self.net(t)


# =====================================================================
# UNET ARCHITECTURE
# =====================================================================
class SelfAttention(nn.Module):
    def __init__(self, channels, num_heads=4):
        super().__init__()
        self.channels = channels
        self.num_heads = num_heads
        self.head_dim = channels // num_heads

        assert channels % num_heads == 0, "channels must be divisible by num_heads"

        self.norm = nn.GroupNorm(8, channels)
        self.qkv = nn.Conv2d(channels, channels * 3, 1)
        self.proj = nn.Conv2d(channels, channels, 1)

    def forward(self, x):
        B, C, H, W = x.shape
        h = self.norm(x)

        qkv = self.qkv(h)
        q, k, v = qkv.chunk(3, dim=1)

        # Reshape for multi-head attention
        q = q.view(B, self.num_heads, self.head_dim, H * W).transpose(2, 3)
        k = k.view(B, self.num_heads, self.head_dim, H * W).transpose(2, 3)
        v = v.view(B, self.num_heads, self.head_dim, H * W).transpose(2, 3)

        # Attention
        scale = self.head_dim ** -0.5
        attn = torch.softmax(q @ k.transpose(-2, -1) * scale, dim=-1)
        h = attn @ v

        # Reshape back
        h = h.transpose(2, 3).contiguous().view(B, C, H, W)
        h = self.proj(h)

        return x + h


class Block(nn.Module):
    def __init__(self, in_ch, out_ch, time_dim, num_groups=8):
        super().__init__()
        self.norm1 = nn.GroupNorm(min(num_groups, in_ch), in_ch)
        self.conv1 = nn.Conv2d(in_ch, out_ch, 3, padding=1)
        self.time_fc = nn.Linear(time_dim, out_ch)
        self.norm2 = nn.GroupNorm(min(num_groups, out_ch), out_ch)
        self.conv2 = nn.Conv2d(out_ch, out_ch, 3, padding=1)
        self.shortcut = (
            nn.Conv2d(in_ch, out_ch, 1) if in_ch != out_ch else nn.Identity()
        )

    def forward(self, x, t_emb):
        h = F.silu(self.norm1(x))
        h = self.conv1(h)
        t_add = self.time_fc(t_emb)[:, :, None, None]
        h = h + t_add
        h = F.silu(self.norm2(h))
        h = self.conv2(h)
        return h + self.shortcut(x)


class Down(nn.Module):
    def __init__(self, in_ch, out_ch, time_dim, use_attn=False):
        super().__init__()
        self.block1 = Block(in_ch, out_ch, time_dim)
        self.block2 = Block(out_ch, out_ch, time_dim)
        self.attn = SelfAttention(out_ch) if use_attn else nn.Identity()
        self.down = nn.Conv2d(out_ch, out_ch, 4, 2, 1)

    def forward(self, x, t_emb):
        x = self.block1(x, t_emb)
        x = self.block2(x, t_emb)
        x = self.attn(x)
        skip = x
        x = self.down(x)
        return x, skip


class Up(nn.Module):
    def __init__(self, in_ch, out_ch, time_dim, use_attn=False):
        super().__init__()
        self.up = nn.ConvTranspose2d(in_ch, out_ch, 4, 2, 1)
        self.block1 = Block(out_ch * 2, out_ch, time_dim)
        self.block2 = Block(out_ch, out_ch, time_dim)
        self.attn = SelfAttention(out_ch) if use_attn else nn.Identity()

    def forward(self, x, skip, t_emb):
        #upsampling with use of skip connections
        x = self.up(x)
        if x.shape[-1] != skip.shape[-1]:
            diff = skip.shape[-1] - x.shape[-1]
            x = F.pad(x, (0, diff, 0, diff))
        x = torch.cat([x, skip], dim=1)
        x = self.block1(x, t_emb)
        x = self.block2(x, t_emb)
        x = self.attn(x)
        return x


class UNet(nn.Module):
    def __init__(
            self,
            img_channels=3,
            base_channels=64,
            channel_mults=(1, 2, 4, 8),
            time_dim=256,
            attn_resolutions=(16,),  # Apply attention at 16x16 resolution
    ):
        super().__init__()
        self.time_dim = time_dim
        self.time_proj = nn.Linear(time_dim, time_dim)
        self.time_mlp = TimeMLP(time_dim)
        #initial Convolution
        self.init_conv = nn.Conv2d(img_channels, base_channels, 3, padding=1)

        ch = base_channels
        self.downs = nn.ModuleList()
        img_res = 96

        #Down Blocks / contracting path
        for mult in channel_mults:
            #base channels multiplied for each layer 64,128,256,512
            out_ch = base_channels * mult 
            #img_res is halved at each downsampling
            img_res = img_res // 2
            use_attn = img_res in attn_resolutions
            self.downs.append(Down(ch, out_ch, time_dim, use_attn=use_attn))
            ch = out_ch

        #Mid Block
        self.mid1 = Block(ch, ch, time_dim)
        self.mid_attn = SelfAttention(ch)
        self.mid2 = Block(ch, ch, time_dim)

        #Up Block / expanding path
        self.ups = nn.ModuleList()
        for mult in reversed(channel_mults):
            #base channels multiplied for each layer, but in reverse 512,256,128,64
            out_ch = base_channels * mult
            #img_res is doubled at each upsampling
            img_res = img_res * 2
            use_attn = img_res in attn_resolutions
            #uses skip connections from down blocks internally
            self.ups.append(Up(ch, out_ch, time_dim, use_attn=use_attn))
            ch = out_ch

        #Final Convolution
        self.final_norm = nn.GroupNorm(8, ch)
        self.final_conv = nn.Conv2d(ch, img_channels, 3, padding=1)

    def forward(self, x, t):
        t_emb = sinusoidal_time_embedding(t, self.time_dim)
        t_emb = F.silu(self.time_proj(t_emb))
        t_emb = self.time_mlp(t_emb)

        x = self.init_conv(x)

        # Downsampling path, save skips
        skips = []
        for down in self.downs:
            x, skip = down(x, t_emb)
            skips.append(skip)

        x = self.mid1(x, t_emb)
        x = self.mid_attn(x)
        x = self.mid2(x, t_emb)

        # Upsampling path, use skips
        for up in self.ups:
            skip = skips.pop()
            x = up(x, skip, t_emb)

        x = F.silu(self.final_norm(x))
        return self.final_conv(x)


# =====================================================================
# DIFFUSION PROCESS
# =====================================================================
class DiffusionForwardProcess:
    def __init__(self, num_timesteps=1000, beta_start=1e-4, beta_end=0.02):
        self.num_timesteps = num_timesteps

        self.betas = torch.linspace(beta_start, beta_end, num_timesteps)
        self.alphas = 1.0 - self.betas
        self.alphas_cumprod = torch.cumprod(self.alphas, dim=0)
        self.alphas_cumprod_prev = torch.cat([torch.tensor([1.0]), self.alphas_cumprod[:-1]])

        self.sqrt_alphas_cumprod = torch.sqrt(self.alphas_cumprod)
        self.sqrt_one_minus_alphas_cumprod = torch.sqrt(1.0 - self.alphas_cumprod)

        self.posterior_variance = (
                self.betas * (1.0 - self.alphas_cumprod_prev) / (1.0 - self.alphas_cumprod)
        )

    def add_noise(self, x_0, noise, t):
        device = x_0.device
        sqrt_alpha_prod = self.sqrt_alphas_cumprod.to(device)[t][:, None, None, None]
        sqrt_one_minus_alpha_prod = self.sqrt_one_minus_alphas_cumprod.to(device)[t][:, None, None, None]
        return sqrt_alpha_prod * x_0 + sqrt_one_minus_alpha_prod * noise


# =====================================================================
# CONFIGURATION
# =====================================================================
class CONFIG:
    num_epochs = 1000
    lr = 2e-4
    num_timesteps = 1000
    batch_size = 16  # Reduced for higher resolution
    img_size = 128  # Increased from 96
    img_channels = 3
    base_channels = 64
    target_classes = [3]  # cats in STL10
    use_ema = True
    sample_every = 50
    gradient_accumulation_steps = 2  # Maintain effective batch size of 32


# =====================================================================
# DATASET
# =====================================================================
def get_dataloader(cfg):
    transform = transforms.Compose([
        transforms.Resize((cfg.img_size, cfg.img_size)),
        transforms.ToTensor(),
        transforms.Normalize([0.5] * 3, [0.5] * 3),  # CRITICAL: [-1, 1]
    ])

    ds = torchvision.datasets.STL10(
        root='./data',
        split='train',
        transform=transform,
        download=True
    )

    labels = [ds[i][1] for i in range(len(ds))]
    indices = [i for i, label in enumerate(labels) if label in cfg.target_classes]
    subset = Subset(ds, indices)

    dl = DataLoader(subset, cfg.batch_size, shuffle=True, num_workers=2, pin_memory=True)
    return dl


def visualize_batch(dataloader):
    images, labels = next(iter(dataloader))
    # Denormalize for display
    images = (images + 1) / 2

    grid = make_grid(images[:16], nrow=4, padding=2)
    plt.figure(figsize=(10, 10))
    plt.imshow(grid.permute(1, 2, 0))
    plt.title("Training Samples")
    plt.axis('off')
    plt.show()



# =====================================================================
# TRAINING
# =====================================================================
def train(cfg, dl):
    device = torch.device('cuda' if torch.cuda.is_available() else 'cpu')
    print(f'Device: {device}')
    print(f'Dataset size: {len(dl.dataset)}')
    print(f'Batch size: {cfg.batch_size}')
    print(f'Total batches per epoch: {len(dl)}\n')

    model = UNet(
        img_channels=cfg.img_channels,
        base_channels=cfg.base_channels,
        channel_mults=(1, 2, 4, 8),
        time_dim=256
    ).to(device)

    optimizer = torch.optim.AdamW(model.parameters(), lr=cfg.lr)
    criterion = nn.MSELoss()
    dfp = DiffusionForwardProcess(num_timesteps=cfg.num_timesteps)

    best_loss = float('inf')
    best_model = None

    for epoch in range(cfg.num_epochs):
        model.train()
        losses = []

        pbar = tqdm(dl, desc=f'Epoch {epoch + 1}/{cfg.num_epochs}')
        for imgs, _ in pbar:
            imgs = imgs.to(device)
            noise = torch.randn_like(imgs).to(device)
            t = torch.randint(0, cfg.num_timesteps, (imgs.shape[0],)).to(device)

            noisy_imgs = dfp.add_noise(imgs, noise, t)

            optimizer.zero_grad()
            noise_pred = model(noisy_imgs, t)
            loss = criterion(noise_pred, noise)

            loss.backward()
            optimizer.step()

            losses.append(loss.item())
            pbar.set_postfix({'loss': f'{loss.item():.4f}'})

        mean_loss = np.mean(losses)
        print(f'Epoch {epoch + 1} | Loss: {mean_loss:.4f}')

        if mean_loss < best_loss:
            best_loss = mean_loss
            best_model = copy.deepcopy(model).cpu()
            torch.cuda.empty_cache()
            print(f'  → New best model (loss: {best_loss:.4f})')

        # Generate samples every 50 epochs
        if (epoch + 1) % cfg.sample_every == 0:
            print(f'\nGenerating samples at epoch {epoch + 1}...')
            model.eval()

            sampler = DiffusionSampler(model, dfp, device)
            with torch.no_grad():
                samples = sampler.ddim_sample(16, cfg.img_channels, cfg.img_size, ddim_steps=50)
                samples = (samples + 1) / 2
                samples = samples.cpu().clamp(0, 1)

            grid = make_grid(samples, nrow=4, padding=2)
            save_image(grid, f'samples_epoch_{epoch + 1:03d}.png')

            # Display inline
            plt.figure(figsize=(10, 10))
            plt.imshow(grid.permute(1, 2, 0))
            plt.title(f'Generated Samples - Epoch {epoch + 1}')
            plt.axis('off')
            plt.show()

            model.train()
            print(f'Samples saved to samples_epoch_{epoch + 1:03d}.png\n')

    print('Training complete!')
    return best_model


# =====================================================================
# SAMPLING
# =====================================================================
class DiffusionSampler:
    def __init__(self, model, diffusion_process, device):
        self.model = model
        self.dfp = diffusion_process
        self.device = device
        self.model.eval()

    @torch.no_grad()
    def ddim_sample(self, batch_size, img_channels, img_size, ddim_steps=50, eta=0.0):
        step_size = self.dfp.num_timesteps // ddim_steps
        timesteps = list(range(0, self.dfp.num_timesteps, step_size))[::-1]

        x = torch.randn(batch_size, img_channels, img_size, img_size).to(self.device)

        for i, t in enumerate(tqdm(timesteps, desc="Sampling")):
            t_batch = torch.full((batch_size,), t, dtype=torch.long).to(self.device)
            predicted_noise = self.model(x, t_batch)

            alpha_cumprod_t = self.dfp.alphas_cumprod[t].to(self.device)

            if i < len(timesteps) - 1:
                t_prev = timesteps[i + 1]
                alpha_cumprod_t_prev = self.dfp.alphas_cumprod[t_prev].to(self.device)
            else:
                alpha_cumprod_t_prev = torch.tensor(1.0).to(self.device)

            pred_x0 = (x - torch.sqrt(1 - alpha_cumprod_t) * predicted_noise) / torch.sqrt(alpha_cumprod_t)
            pred_x0 = torch.clamp(pred_x0, -1.0, 1.0)

            dir_xt = torch.sqrt(1 - alpha_cumprod_t_prev) * predicted_noise

            if eta > 0 and i < len(timesteps) - 1:
                noise = torch.randn_like(x)
                sigma_t = eta * torch.sqrt(
                    (1 - alpha_cumprod_t_prev) / (1 - alpha_cumprod_t) * (1 - alpha_cumprod_t / alpha_cumprod_t_prev))
            else:
                noise = torch.zeros_like(x)
                sigma_t = 0

            x = torch.sqrt(alpha_cumprod_t_prev) * pred_x0 + dir_xt + sigma_t * noise

        return x


def generate_and_display(model, num_images=16, img_size=96, ddim_steps=50):
    device = torch.device('cuda' if torch.cuda.is_available() else 'cpu')
    model = model.to(device)

    dfp = DiffusionForwardProcess(num_timesteps=1000)
    sampler = DiffusionSampler(model, dfp, device)

    images = sampler.ddim_sample(num_images, img_channels=3, img_size=img_size, ddim_steps=ddim_steps)

    # Denormalize from [-1, 1] to [0, 1]
    images = (images + 1) / 2
    images = images.cpu().clamp(0, 1)

    grid = make_grid(images, nrow=int(np.sqrt(num_images)), padding=2)

    plt.figure(figsize=(12, 12))
    plt.imshow(grid.permute(1, 2, 0))
    plt.title("Generated Images")
    plt.axis('off')
    plt.show()

    return images


def save_generated_images(model, save_path='generated.png', num_images=16, img_size=96, ddim_steps=50):
    device = torch.device('cuda' if torch.cuda.is_available() else 'cpu')
    model = model.to(device)

    dfp = DiffusionForwardProcess(num_timesteps=1000)
    sampler = DiffusionSampler(model, dfp, device)

    images = sampler.ddim_sample(num_images, img_channels=3, img_size=img_size, ddim_steps=ddim_steps)
    images = (images + 1) / 2
    images = images.cpu().clamp(0, 1)

    grid = make_grid(images, nrow=int(np.sqrt(num_images)), padding=2)
    save_image(grid, save_path)
    print(f'Saved to {save_path}')

    return images


# =====================================================================
# EXECUTION
# =====================================================================
# Initialize config
cfg = CONFIG()

# Load data
dataloader = get_dataloader(cfg)

# Visualize training data
visualize_batch(dataloader)

# Train model
trained_model = train(cfg, dataloader)

# Save model
torch.save(trained_model, 'ddpm_cats_model.pth')

# Generate images
generated = generate_and_display(trained_model, num_images=16, img_size=96, ddim_steps=50)

# Save generated images
save_generated_images(trained_model, 'generated_cats.png', num_images=16, img_size=96, ddim_steps=50)
```
